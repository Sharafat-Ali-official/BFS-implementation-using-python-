#Part 1 Function with list 

use of map fucntion provide us easy way to create a new list using function parameters

```python
def func1(listparam):
    list_return =list(map(lambda x: x*x, listparam))
    return list_return

list1 = [1,2,3,4,5]
func1(list1)
```

#Part 3: student class using OOP concepts -> class and object

```python
class student:
    def __init__(self,name,roll_no,semester):
        self.name=name
        self.roll_no=roll_no
        self.semester=semester

    def display_information(self):
        print(f" name: {self.name}, \n roll_no: {self.roll_no}, \n semester: {self.semester}\n")


student1=student("John",101,5)
student2=student("Alice",102,6)


#displaying information using method calling function 
student1.display_information()
student2.display_information()
```

PART 4 Breadth First Search Algorithm  Implementation 

tree node created 

bfs fucntion use to create a tree and use deque for double sided queue

popleft function is used to pop the queue element from left 

```python
from collections import deque as dq

class treenode:
    def __init__(self,value):
        self.value=value
        self.children=[]

def bfs(root):
    if root is None:
        return

    queue=dq([root])
    resultant=[]
    while queue:
        current_node=queue.popleft()
        print(f" \n node poped from queue: {current_node.value} \n")
        resultant.append(current_node.value)

        print(f" \n queue after poping: {[x.value for x in queue]} \n")

        # children of current node are added to the queue
        for child in current_node.children:
            queue.append(child)

        print(f" \n queue after adding children: {[x.value for x in queue]} \n")

    return resultant


root=treenode('A')


root.children = [B, C, D]
B=treenode('B')
C=treenode('C')
D=treenode('D')

E=treenode('E')
F=treenode('F')

B.children = [E, F]

G=treenode('G')
H=treenode('H')
I=treenode('I')

C.children = [G, H, I]

J=treenode('J')

D.children = [J]

K=treenode('K')
L=treenode('L')

G.children = [K, L]




bfs_result=bfs(root)
print(f" \n bfs result: {bfs_result} \n")
```
